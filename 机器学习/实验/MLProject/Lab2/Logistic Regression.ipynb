```python
import sklearn
import matplotlib.pyplot as plt
from sklearn import model_selection
from sklearn import metrics
from sklearn import linear_model
import pandas as pd
import numpy as np
```

```python
data_path = 'Logistic Dataset.xlsx'
data = pd.read_excel(data_path)
data
```

```python
x_0 = data['密度'].to_numpy().reshape(-1,1)
x_1 = data['含糖率'].to_numpy().reshape(-1,1)
y = data['好瓜'].to_numpy().reshape(-1,1)
x = np.hstack((x_0,x_1))
```

```python
x_train,x_test,y_train,y_test=model_selection.train_test_split(x,y,test_size=0.1)
x_train.shape
```

```python
logistic = linear_model.LogisticRegression()
logistic.fit(x_train, y_train)
```

```python
y_train_pre = logistic.predict(x_train)
plt.plot([i for i in range(len(y_train_pre))], y_train_pre)
plt.plot([i for i in range(len(y_train))], y_train)
plt.show()
```

```python
y_test_pre = logistic.predict(x_test)
y_test,y_test_pre
```

```python
acc = np.mean(y_test_pre==y_test)
acc_train = np.mean(y_train_pre==y_train)
acc_train,acc
```
