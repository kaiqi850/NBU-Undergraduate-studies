```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from sklearn import model_selection
from sklearn import metrics
from sklearn import linear_model
from tqdm import tqdm
```

```python
# 线性归回数据读入
data_path = 'Linear Dataset.xlsx'
data = pd.read_excel(data_path)
data
```

```python
x_0 = data['密度'].to_numpy().reshape(-1,1)
x_1 = data['含糖量'].to_numpy().reshape(-1,1)
y = data['成熟度'].to_numpy().reshape(-1,1)
x = np.hstack((x_0,x_1))
x.shape
```

```python
x_train,x_test,y_train,y_test=model_selection.train_test_split(x,y,test_size=0.1)
```

```python
linear = linear_model.LinearRegression()
linear.fit(x_train, y_train)
```

```python
# 显示模型输出与真实结果，观察拟合程度
y_train_pre = linear.predict(x_train)
plt.plot([i for i in range(len(y_train_pre))], y_train_pre)
plt.plot([i for i in range(len(y_train))], y_train)
plt.legend(['model label','true label'], loc='best')
plt.xlabel('id')
plt.ylabel('value')
plt.title('Train result')
plt.show()
```

```python
# 测试集上的均方误差
y_test_pre=linear.predict(x_test)
SQ_ERR = metrics.mean_squared_error(y_test_pre, y_test)
print(SQ_ERR, linear.coef_)
```
