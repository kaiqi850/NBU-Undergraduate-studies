```python
import numpy as np
import pandas as pd
from tqdm import tqdm
import matplotlib.pyplot as plt
from sklearn import model_selection
from sklearn import metrics
```

```python
data_path = 'Linear Dataset.xlsx'
data = pd.read_excel(data_path)
x_0 = data['密度'].to_numpy().reshape(-1,1)
x_1 = data['含糖量'].to_numpy().reshape(-1,1)
y = data['成熟度'].to_numpy().reshape(-1,1)
x = np.hstack((x_0,x_1))
x = np.hstack((x, np.ones([y.shape[0],1])))
x
```

```python
x_train,x_test,y_train,y_test=model_selection.train_test_split(x,y,test_size=0.1)
```

```python
# 均方误差
def LossFunction(theta,feature,label):
    return np.sum((np.dot(feature,theta)-label)**2)/(2*feature.shape[0])
# 线性回归模型训练
def linear_train(theta, feature, label, iter=100, lr=0.001, loss_list=None):
    num=feature.shape[0]
    for i in tqdm(range(iter)):
        # 梯度计算
        dL=np.dot(feature.transpose(),(np.dot(feature,theta)-label))/num
        theta=theta-lr*dL # 沿梯度下降
        Loss=LossFunction(theta,feature,label)
        loss_list.append(Loss)
# 根据模型参数进行预测
def predict(theta, feature):
    num=feature.shape[0]
    pre=[]
    for i in range(num):
        pre.append(np.dot(feature[i],theta))
    return np.array(pre)
```

```python
params=np.random.rand(3,1)
loss_list=[]
linear_train(params, x_train, y_train, iter=12, lr=10,loss_list=loss_list)
y_train_pre=predict(params,x_train)
y_test_pre=predict(params,x_test)
```

```python
plt.plot([i for i in range(len(loss_list))], loss_list)
plt.legend(['Loss with lr = 10'], loc='best')
plt.xlabel('iterations')
plt.ylabel('Loss')
plt.title('Training process')
```

```python
plt.plot([i for i in range(len(y_train_pre))], y_train_pre)
plt.plot([i for i in range(len(y_train))], y_train)
plt.legend(['model label','true label'], loc='best')
plt.xlabel('id')
plt.ylabel('value')
plt.title('Train result')
plt.show()
```

```python
SQ_ERR = metrics.mean_squared_error(y_test_pre, y_test)
SQ_ERR
```
