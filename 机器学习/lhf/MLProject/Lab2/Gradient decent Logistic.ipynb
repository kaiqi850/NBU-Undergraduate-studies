```python
import numpy as np
import pandas as pd
from tqdm import tqdm
import matplotlib.pyplot as plt
from sklearn import model_selection
from sklearn import metrics
```

```python
data_path = 'Logistic Dataset.xlsx'
data = pd.read_excel(data_path)
x_0 = data['密度'].to_numpy().reshape(-1,1)
x_1 = data['含糖率'].to_numpy().reshape(-1,1)
y = data['好瓜'].to_numpy().reshape(-1,1)
x = np.hstack((x_0,x_1))
x = np.hstack((x, np.ones([y.shape[0],1])))
x
```

```python
x_train,x_test,y_train,y_test=model_selection.train_test_split(x,y,test_size=0.1)
x_train.shape
```

```python
# sigmoid函数，值域(0,1)
def sigmoid(z):
    return 1/(1+np.exp(-z))
# 逻辑回归损失函数（交叉熵）
def LossFunction(theta,feature,label):
    num=feature.shape[0]
    return -1/num*(np.dot(label.transpose(),np.log(sigmoid(np.dot(feature,theta))))+np.dot((1-label).transpose(),np.log(1-sigmoid(np.dot(feature,theta)))))
# 逻辑回归模型训练
def logistic_train(theta, feature, label, iter=100, lr=0.001, loss_list=None):
    num=feature.shape[0]
    for i in tqdm(range(iter)):
        # 梯度计算
        dL=np.dot(feature.transpose(),(sigmoid(np.dot(feature,theta))-label))/num
        theta=theta-lr*dL  #沿着梯度下降
        Loss=LossFunction(theta,feature,label).item()
        loss_list.append(Loss)
# 根据模型参数进行预测
def predict(theta, feature):
    pre=np.zeros([feature.shape[0],1])
    for i,val in enumerate(np.dot(feature,theta)):
        if sigmoid(val)>=0.5:
            pre[i]=1
        else:
            pre[i]=0
    return pre
```

```python
params=np.random.rand(3,1)
loss_list=[]
logistic_train(params, x_train, y_train, iter=100000, lr=0.01,loss_list=loss_list)
y_train_pre=predict(params,x_train)
y_test_pre=predict(params,x_test)
loss_list
```

```python
plt.plot([i for i in range(len(loss_list))], loss_list)
plt.legend(['Loss'], loc='best')
plt.xlabel('iterations')
plt.ylabel('Loss')
plt.title('Training process')
```

```python
plt.plot([i for i in range(len(y_train_pre))], y_train_pre)
plt.plot([i for i in range(len(y_train))], y_train)
```

```python
acc = np.mean(y_test_pre==y_test)
acc
```
