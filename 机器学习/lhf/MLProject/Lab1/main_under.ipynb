```python
# 相关库导入
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
```

```python
rnd  = np.random.RandomState(42)
# 训练集，样本数100
x_train = rnd.uniform(-3, 3, size=100)
y_train = np.sin(x_train) + rnd.normal(size=len(x_train)) / 3
# 测试集，样本数100
x_test = rnd.uniform(-3, 3, size=100)
y_test = np.sin(x_test) + rnd.normal(size=len(x_test)) / 3

x_train.shape, y_train.shape
```

```python
# 训练集、测试集数据绘制
fig, ax1 = plt.subplots(1)
# 训练集数据，图像为绿色，标签为train dataset
ax1.plot(x_train, y_train, 'o', c='g', label='train dataset')
# 测试集数据，图像为蓝色，标签为test dataset
ax1.plot(x_test, y_test, 'o', c='b', label='test dataset')
# 将标签放置于最佳位置
ax1.legend(loc='best')
# 设置其余图像选项，包括横纵坐标标签、图像标题
ax1.set_ylabel('y')
ax1.set_xlabel('x')
ax1.set_title("Train Dataset Display")
# 图像尺寸调整
plt.tight_layout()
plt.show()
```

```python
# 数据维度改变：变为shape为[n,f]，其中n为样本数，f为特征数，本例为1
x_train_reshape = x_train.reshape(-1, 1)
x_test_reshape = x_test.reshape(-1, 1)
x_train_reshape.shape, x_test_reshape.shape
```

```python
# 将数据集进行特征转换，调整为多项式特征，其中度为4的多项式具有5个特征值，故shape为[n,5]
poly = PolynomialFeatures(degree=2)
x_train_poly = poly.fit_transform(x_train_reshape)
x_train_poly.shape
```

```python
# 模型实例化：线性回归模型
lr = LinearRegression()
# 在训练集上训练数据
lr.fit(x_train_poly, y_train)
y_predict = lr.predict(x_train_poly)
```

```python
# 在测试集上测试模型
x_test_poly = poly.fit_transform(x_test_reshape)
y_test_predict = lr.predict(x_test_poly)
```

```python
# 绘制图像：训练集的拟合情况
fig, ax1 = plt.subplots(1)
ax1.plot(x_train, y_train, 'o', c='k',label='train point dataset')
ax1.plot(np.sort(x_train), y_predict[np.argsort(x_train)], linewidth=2, c='g', label='train regression')
ax1.legend(loc='best')
ax1.set_xlabel('x')
ax1.set_ylabel('y')
ax1.set_title('Train Result Display')
plt.tight_layout()
plt.show()
plt.close()
```

```python
fig, ax1 = plt.subplots(1)
ax1.plot(x_test, y_test, 'o', c='r', label='test point dataset')
ax1.plot(np.sort(x_test), y_test_predict[np.argsort(x_test)], linewidth=2, c='b', label='test regression')
ax1.legend(loc='best')
ax1.set_xlabel('x')
ax1.set_ylabel('y')
ax1.set_title('Test Result Display')
plt.tight_layout()
plt.show()
plt.close()
```
