```python
import matplotlib.pyplot as plt
import os
import numpy as np
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
```

```python
def load_image(file_path):
    data = plt.imread(file_path)
    return data

def show_image(image):
    plt.imshow(image, cmap='gray')
    plt.show()

def save_image(name, image):
    plt.imsave(name, image, cmap='gray')
```

```python
def presolve_data():
    root = r'./archive/data'
    filenames = os.listdir(root)
    for i, filename in enumerate(filenames):
        image = load_image(os.path.join(root, filename))
        save_name = str(i) + '.png'
        save_image(os.path.join(r'./presolve_data/', save_name), image)
```

```python
def get_feature():
    root = r'./archive/data'
    filenames = os.listdir(root)
    all_images_flattened = np.empty((165, 320 * 243), dtype=np.float32)
    for i, filename in enumerate(filenames):
        image = load_image(os.path.join(root, filename))
        image = image.flatten()
        all_images_flattened[i, :] = image
    return all_images_flattened
```

```python
def pca_solve():
    X = get_feature()
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)
    print(X_scaled.shape)
    pca = PCA(n_components=100)
    X_reduced = pca.fit_transform(X_scaled)
    print(X_reduced.shape)
    n_images_to_plot = 5
    fig, axes = plt.subplots(2, n_images_to_plot, figsize=(8, 3), subplot_kw={'xticks': (), 'yticks': ()})
    for i, ax in enumerate(axes.flat):
        # 原始图像
        ax.imshow(X[i, :].reshape(243, 320), cmap='gray')
        ax.set_title('Before')
        ax.set_xlabel('image %d' % i)
    for i, (ax, orig) in enumerate(zip(axes.flat[n_images_to_plot:], X)):
        X_reconstructed = pca.inverse_transform(X_reduced[i:i+1])
        image_reconstructed = X_reconstructed.reshape(243, 320)
        # 显示重构后的图像
        ax.imshow(image_reconstructed, cmap='gray')
        ax.set_title('After')
        ax.set_xlabel('image %d' % i)
```

```python
pca_solve()
```
